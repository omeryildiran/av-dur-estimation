```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from ipywidgets import interact, FloatSlider, Layout
import warnings
warnings.filterwarnings('ignore')

def plot_gaussian_average(mu1, sigma1, mu2, sigma2, w1):
    w2 = 1.0 - w1
    x = np.linspace(-10, 10, 1000)

    g1 = norm.pdf(x, mu1, sigma1)
    g2 = norm.pdf(x, mu2, sigma2)
    mixture = w1 * g1 + w2 * g2

    fig, ax = plt.subplots(figsize=(9, 5))

    ax.plot(x, g1, color='steelblue', lw=2, label=f'G1: μ={mu1:.1f}, σ={sigma1:.1f}  (w={w1:.2f})')
    ax.fill_between(x, g1, alpha=0.15, color='steelblue')

    ax.plot(x, g2, color='tomato', lw=2, label=f'G2: μ={mu2:.1f}, σ={sigma2:.1f}  (w={w2:.2f})')
    ax.fill_between(x, g2, alpha=0.15, color='tomato')

    ax.plot(x, mixture, color='purple', lw=2.5, ls='--', label='Weighted mixture')
    ax.fill_between(x, mixture, alpha=0.10, color='purple')

    ax.axvline(mu1, color='steelblue', lw=1, ls=':')
    ax.axvline(mu2, color='tomato',    lw=1, ls=':')

    mix_mean = w1 * mu1 + w2 * mu2
    ax.axvline(mix_mean, color='purple', lw=1.2, ls=':')
    ax.text(mix_mean, ax.get_ylim()[1]*0.95, f'  mix μ={mix_mean:.2f}',
            color='purple', fontsize=9, va='top')

    ax.set_xlabel('x', fontsize=12)
    ax.set_ylabel('Probability density', fontsize=12)
    ax.set_title('Weighted Gaussian Mixture', fontsize=13)
    ax.legend(fontsize=10)
    ax.set_xlim(-10, 10)
    ax.set_ylim(bottom=0)
    ax.tick_params(axis='x', rotation=90)
    plt.tight_layout()
    plt.savefig('gaussian_mixture.svg', format='svg')
    plt.show()

slider_kw = dict(style={'description_width': '120px'},
                 layout=Layout(width='420px'))

interact(
    plot_gaussian_average,
    mu1    = FloatSlider(value=-2.0, min=-8.0, max=8.0,  step=0.1, description='G1 mean (μ₁)',  **slider_kw),
    sigma1 = FloatSlider(value=1.0,  min=0.1,  max=4.0,  step=0.1, description='G1 std  (σ₁)',  **slider_kw),
    mu2    = FloatSlider(value=2.0,  min=-8.0, max=8.0,  step=0.1, description='G2 mean (μ₂)',  **slider_kw),
    sigma2 = FloatSlider(value=1.5,  min=0.1,  max=4.0,  step=0.1, description='G2 std  (σ₂)',  **slider_kw),
    w1     = FloatSlider(value=0.8,  min=0.0,  max=1.0,  step=0.01,description='Weight G1 (w₁)', **slider_kw),
);
```
